```python
print("Dataset Shape:", df.shape)
print("\nContent Types:")
print(df['type'].value_counts())

print("\nRatings:")
print(df['rating'].value_counts())

print("\nRelease Year Range:")
print(df['release_year'].min(), df['release_year'].max())

print("\nDate Added Sample:")
print(df['date_added'].head(10))

print("\nDuration Sample:")
print(df['duration'].value_counts().head(15))
```

```python
# Find records where rating contains a duration
incorrect_ratings = df[
    df['rating'].str.contains(r'^\d+\s+min$', na=False)
]

print(incorrect_ratings[
    ['show_id', 'type', 'title', 'rating', 'duration']
])
```

```python
df['date_added'] = pd.to_datetime(
    df['date_added'].str.strip(),
    errors='coerce'
)

print("Earliest Date Added:", df['date_added'].min())
print("Latest Date Added:", df['date_added'].max())

print("\nMissing Dates:", df['date_added'].isnull().sum())
```

```python
# Identify duration values incorrectly stored in rating
mask = df['rating'].str.contains(r'^\d+\s+min$', na=False)

# Move those values into duration
df.loc[mask, 'duration'] = df.loc[mask, 'rating']

# Clear the incorrect rating values
df.loc[mask, 'rating'] = None

# Verify
print(df.loc[mask, ['title', 'rating', 'duration']])
```

```python
# Fill missing categorical information
columns = ['director', 'cast', 'country', 'rating']

for col in columns:
    df[col] = df[col].fillna('Unknown')

# Check missing values
print(df.isnull().sum())
```

```python
df['date_added_year'] = df['date_added'].dt.year

df['date_added_month'] = df['date_added'].dt.month_name()

df['date_added_month_num'] = df['date_added'].dt.month
```

```python
df['duration_minutes'] = pd.to_numeric(
    df['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

df['duration_seasons'] = pd.to_numeric(
    df['duration'].str.extract(r'(\d+)\s+Season')[0],
    errors='coerce'
)
```

```python
print(df.shape)
print(df.isnull().sum())
print(df[['title', 'type', 'duration',
          'duration_minutes', 'duration_seasons']].head(10))
```

```python
# Extract numerical duration
df['duration_value'] = pd.to_numeric(
    df['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Separate movie duration and TV show seasons
df['duration_minutes'] = df['duration_value'].where(
    df['type'] == 'Movie'
)

df['duration_seasons'] = df['duration_value'].where(
    df['type'] == 'TV Show'
)

# Remove temporary column
df.drop(columns=['duration_value'], inplace=True)
```

```python
print(df.groupby('type')[[
    'duration_minutes',
    'duration_seasons'
]].count())

print("\nMovie duration:")
print(df[df['type'] == 'Movie']['duration_minutes'].describe())

print("\nTV show seasons:")
print(df[df['type'] == 'TV Show']['duration_seasons'].describe())
```

```python
# Number of unique genre combinations
print("Unique genre combinations:", df['listed_in'].nunique())

# Most frequent genre combinations
print("\nTop 15 genre combinations:")
print(df['listed_in'].value_counts().head(15))

# Number of individual genres
genres = df['listed_in'].str.split(',').explode().str.strip()

print("\nUnique individual genres:", genres.nunique())

print("\nTop 15 individual genres:")
print(genres.value_counts().head(15))
```

```python
genre_df = df[['show_id', 'type', 'title', 'listed_in']].copy()

genre_df['genre'] = genre_df['listed_in'].str.split(',')

genre_df = genre_df.explode('genre')

genre_df['genre'] = genre_df['genre'].str.strip()

genre_df = genre_df.drop(columns=['listed_in'])

genre_df.head(10)
```

```python
print("Genre table shape:", genre_df.shape)

print("\nTotal genre records:", len(genre_df))

print("\nTop 15 genres:")
print(genre_df['genre'].value_counts().head(15))

print("\nDuplicate title-genre combinations:",
      genre_df.duplicated(subset=['show_id', 'genre']).sum())
```

```python
country_df = df[['show_id', 'type', 'title', 'country']].copy()

country_df['country'] = country_df['country'].str.split(',')

country_df = country_df.explode('country')

country_df['country'] = country_df['country'].str.strip()

print("Country table shape:", country_df.shape)

print("\nTop 15 countries:")
print(country_df['country'].value_counts().head(15))
```

```python
main_df = df[
    [
        'show_id',
        'type',
        'title',
        'director',
        'cast',
        'country',
        'date_added',
        'date_added_year',
        'date_added_month',
        'date_added_month_num',
        'release_year',
        'rating',
        'duration',
        'duration_minutes',
        'duration_seasons',
        'listed_in',
        'description'
    ]
].copy()
```

```python
print("Main table:", main_df.shape)
print("Genre table:", genre_df.shape)
print("Country table:", country_df.shape)

print("\nMissing values in main table:")
print(main_df.isnull().sum())

print("\nSample main table:")
display(main_df.head())

print("\nSample genre table:")
display(genre_df.head())

print("\nSample country table:")
display(country_df.head())
```

```python
main_df.to_csv('netflix_main.csv', index=False)
genre_df.to_csv('netflix_genres.csv', index=False)
country_df.to_csv('netflix_countries.csv', index=False)

print("All three datasets exported successfully!")
```
